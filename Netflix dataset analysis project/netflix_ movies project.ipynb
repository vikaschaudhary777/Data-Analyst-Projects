```python
# Fill country & rating
df['country'].fillna('Unknown', inplace=True)
df['rating'].fillna('Unknown', inplace=True)

# Drop small missing cases
df.dropna(subset=['date_added', 'duration'], inplace=True)
```

```python
df['date_added'] = df['date_added'].str.strip()

# Convert to datetime, allowing mixed formats
df['date_added'] = pd.to_datetime(df['date_added'], format='mixed', errors='coerce')
```

```python
df['year_added'] = df['date_added'].dt.year
df['month_added'] = df['date_added'].dt.month_name()
```

```python
df.columns = df.columns.str.strip().str.lower().str.replace(' ', '_')
```

```python
# Strip leading/trailing spaces from all string columns
for col in df.select_dtypes(include='object').columns:
    df[col] = df[col].str.strip()
```

# Next Step – EDA

1. Movies & TV Shows released by month

```python
import plotly.express as px

# Count shows by month
month_count = df['month_added'].value_counts().reindex([
    'January', 'February', 'March', 'April', 'May', 'June',
    'July', 'August', 'September', 'October', 'November', 'December'
])

# Reset index so we can use color easily
month_count_df = month_count.reset_index()
month_count_df.columns = ['Month', 'Count']

# Plot with a gradient color scale based on Count
fig = px.bar(
    month_count_df,
    x='Month',
    y='Count',
    color='Count',  # Color intensity based on count
    color_continuous_scale='Plasma',  # You can try 'Viridis', 'Plasma', 'Cividis'
    title='Movies & TV Shows Added by Month on Netflix',
    labels={'Count': 'Number of Movies & TV Shows'}
)

fig.show()
```

2. Unique ratings & most common rating

```python
rating_count = df['rating'].value_counts()

fig = px.bar(
    x=rating_count.index,
    y=rating_count.values,
    labels={'x': 'Rating', 'y': 'Count'},
    title='Distribution of Ratings'
)
fig.show()
```

3. Year-on-Year content addition by type

```python
year_type = df.groupby(['year_added', 'type']).size().reset_index(name='count')

fig = px.bar(
    year_type,
    x='year_added',
    y='count',
    color='type',
    barmode='group',
    title='Year-on-Year Netflix Additions by Type'
)
fig.show()
```

4. Salman, Shah Rukh, and Akshay Kumar – all movies

```python
# Salman Khan movies
salman_movies = df[
    df['cast'].notna() &
    df['cast'].str.contains('Salman Khan', case=False)
]

# Shah Rukh Khan movies
srk_movies = df[
    df['cast'].notna() &
    df['cast'].str.contains('Shah Rukh Khan', case=False)
]

# Akshay Kumar movies
akshay_movies = df[
    df['cast'].notna() &
    df['cast'].str.contains('Akshay Kumar', case=False)
]

print("Salman Khan Movies:", len(salman_movies))
print("Shah Rukh Khan Movies:", len(srk_movies))
print("Akshay Kumar Movies:", len(akshay_movies))
```

5. Movies & Series with Maximum Time Length

```python
# For movies
max_movie = df[df['type'] == 'Movie'].sort_values(
    by='duration', ascending=False).head(1)

# For TV Shows
max_show = df[df['type'] == 'TV Show'].sort_values(
    by='duration', ascending=False).head(1)

print("Longest Movie:\n", max_movie[['title', 'duration']])
print("Longest TV Show:\n", max_show[['title', 'duration']])
```

6. Akshay Kumar all comedy movies, Shah Rukh movies with Kajol, Salman–Akshay movies

```python
from tabulate import tabulate

# Akshay Kumar comedy movies
akshay_comedy = df[
    df['cast'].notna() &
    df['cast'].str.contains('Akshay Kumar', case=False) &
    df['listed_in'].str.contains('Comedy', case=False)
]
print(f"🎭 Akshay Kumar Comedy Movies ({len(akshay_comedy)} found):")
print(tabulate(
    akshay_comedy[['title', 'type', 'release_year', 'listed_in']].head(10),
    headers='keys',
    tablefmt='fancy_grid'
), "\n")

# Shah Rukh Khan movies with Kajol
srk_kajol = df[
    df['cast'].notna() &
    df['cast'].str.contains('Shah Rukh Khan', case=False) &
    df['cast'].str.contains('Kajol', case=False)
]
print(f"💖 Shah Rukh Khan with Kajol Movies ({len(srk_kajol)} found):")
print(tabulate(
    srk_kajol[['title', 'type', 'release_year', 'listed_in']].head(10),
    headers='keys',
    tablefmt='fancy_grid'
), "\n")

# Salman Khan with Akshay Kumar movies
salman_akshay = df[
    df['cast'].notna() &
    df['cast'].str.contains('Salman Khan', case=False) &
    df['cast'].str.contains('Akshay Kumar', case=False)
]
print(f"🤝 Salman Khan & Akshay Kumar Movies ({len(salman_akshay)} found):")
print(tabulate(
    salman_akshay[['title', 'type', 'release_year', 'listed_in']].head(10),
    headers='keys',
    tablefmt='fancy_grid'
))
```

7. Director who has made the most TV Shows

```python
top_tv_director = (
    df[df['type'] == 'TV Show']
    .groupby('director')['title']
    .count()
    .sort_values(ascending=False)
    .head(1)
)
print(top_tv_director)
```

8. Actors/Actresses with most number of Movies

```python
from collections import Counter
from tabulate import tabulate

# Filter only movies with cast info
movies_with_cast = df[df['type'] == 'Movie']['cast'].dropna()

# Create actor list
all_actors = []
for cast in movies_with_cast:
    all_actors.extend([actor.strip() for actor in cast.split(',')])

# Count most common actors
actor_counts = Counter(all_actors).most_common(10)

# Convert to DataFrame-like list for tabulate
table_data = [(i+1, actor, count) for i, (actor, count) in enumerate(actor_counts)]

# Display as table
print("🎬 Top 10 Most Frequent Actors in Movies")
print(tabulate(table_data, headers=["Rank", "Actor", "No. of Movies"], tablefmt="fancy_grid"))
```

 9. Genre with most Movies & TV Shows

```python
# Split 'listed_in' into individual genres and count
genre_counts = (
    df['listed_in']
    .dropna()
    .str.split(',')
    .explode()
    .str.strip()
    .value_counts()
)

# Prepare table data
table_data = [(i+1, genre, count) for i, (genre, count) in enumerate(genre_counts.items())]

# Display as optional table
print("📊 Genre with Most Movies & TV Shows")
print(tabulate(table_data, headers=["Rank", "Genre", "Count"], tablefmt="fancy_grid"))
```
